```python
 # Import Libraries

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Car_Price_Data.csv")
df.info()
```

```python
 # Clean Price Column

df['Price'].unique().tolist()
df = df[df['Price'] != 'Ask For Price']
df['Price'] = df['Price'].str.replace(',', '')
df['Price'] = df['Price'].astype(int)
```

```python
df['Price'].unique().tolist()
```

```python
df.info()
```

```python
 # Clean Year Column

df['year'].unique().tolist()
df = df[df['year'].str.isnumeric()]
df['year'] = df['year'].astype(int)
```

```python
df['year'].unique().tolist()
```

```python
 # Clean kms column

df['kms_driven'].unique().tolist()
df = df.dropna(subset=['kms_driven'])
df['kms_driven'] = df['kms_driven'].str.replace(' kms', '')
df['kms_driven'] = df['kms_driven'].str.replace(',', '')
df = df[df['kms_driven'].str.isnumeric()]
df['kms_driven'] = df['kms_driven'].astype(int)
```

```python
df['kms_driven'].unique().tolist()
```

```python
 # Clean Fuel Column

df['fuel_type'].unique().tolist()
df = df.dropna(subset=["fuel_type"])
df.info()
```

```python
df['fuel_type'].unique().tolist()
```

```python
# STEP 6: Remove Whitespace and Text Anomalies

df["name"] = df["name"].str.strip()
df["company"] = df["company"].str.strip()
df["fuel_type"] = df["fuel_type"].str.strip()
```

```python
df
```

```python
# STEP 7: Reset the Index and Save Clean Data

df = df.reset_index(drop=True)
```

```python
df
```

```python
df.duplicated().value_counts()
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df.duplicated().value_counts()
```

```python
for i in df.index:
    name = df.loc[i,"name"]
    company = df.loc[i,"company"]
    name = name.replace(company + " ","")
    name = ' '.join(name.split()[:3])
    df.loc[i,"name"] = name
```

```python
df
```

```python
# Save the polished data to a new CSV file
df.to_csv("Cleaned_Car_Price_Data.csv", index=False)
```
